# Aula Semana 01: Guia Definitivo da Metodologia CRISP-DM em Ciência de Dados

## 1. Introdução à Metodologia CRISP-DM

No cenário corporativo moderno, grande parte dos projetos de Inteligência Artificial e Ciência de Dados falha não por limitações dos algoritmos, mas por **falta de alinhamento com os objetivos estratégicos do negócio, má gestão do ciclo de vida dos dados e ausência de uma metodologia estruturada**.

Para superar esses desafios, em 1996 um consórcio liderado por empresas como Daimler-Benz, SPSS e NCR desenvolveu o **CRISP-DM** (*Cross-Industry Standard Process for Data Mining*).

### Diferença entre Desenvolvimento de Software Tradicional e Ciência de Dados

| Aspecto | Desenvolvimento Ágil Tradicional (Scrum/Kanban) | Ciência de Dados (CRISP-DM) |
| :--- | :--- | :--- |
| **Natureza do Trabalho** | Engenharia determinística (requisito -> código -> funcionalidade) | Experimental, estocástico e baseado em validação de hipóteses |
| **Requisitos** | Funcionais e não-funcionais bem especificados | Objetivos de negócio sujeitos a descobertas exploratórias |
| **Métrica de Sucesso** | Software funcionando sem bugs conforme especificação | Desempenho estatístico do modelo e Retorno Financeiro (ROI) |
| **Ciclo de Vida** | Linear/Iterativo em Sprints | Não-linear, cíclico e altamente retroalimentado |

```text
│   └───────────┬────────────┘         └───────────┬────────────┘        │
│               │                                  │                     │

│   ┌────────────────────────┐         ┌────────────────────────┐        │
│   │ 6. Deployment          │         │ 3. Data                │        │
│   │    (Implantação)       │         │    Preparation         │        │

```

### Fase 1: Business Understanding (Entendimento do Negócio)
Nesta fase inicial, definem-se as necessidades reais da organização:
1. **Determinação dos Objetivos do Negócio**: O que a empresa quer alcançar? (ex: Reduzir a inadimplência de crédito em 15%).
2. **Avaliação da Situação Atual**: Recursos disponíveis, riscos envolvidos, requisitos regulatórios (LGPD/BACEN) e restrições técnicas.
3. **Determinação das Metas de Ciência de Dados**: Tradução das metas de negócio em métricas técnicas estatísticas (ex: Construir modelo de classificação binária com Recall da classe inadimplente >= 80%).
4. **Produção do Plano de Projeto**: Escolha das ferramentas, cronograma e estratégia de amostragem.

#### Mapeamento de Custos Assimétricos
Em problemas de concessão de crédito ou detecção de fraudes, comete-se dois tipos de erros de classificação que possuem impactos econômicos completamente distintos:
- **Falso Negativo (FN - Erro Tipo II)**: O modelo classifica um cliente inadimplente como "bom pagador", concedendo o empréstimo. *Consequência: Perda total do capital emprestado (ex: R$ 10.000,00 por contrato).*
- **Falso Positivo (FP - Erro Tipo I)**: O modelo classifica um cliente bom pagador como "inadimplente", recusando o empréstimo. *Consequência: Perda da margem de lucro de juros que seria arrecadada (ex: R$ 600,00 por proposta).*

Por essa razão, métricas como **Acurácia simples são enganosas**, e o projeto deve ser otimizado para o **Recall** da classe de risco.

---

### Fase 2: Data Understanding (Entendimento dos Dados)
Nesta fase, realiza-se a imersão nos dados brutos:
- **Coleta de Dados**: Extração de bancos relacionais (SQL), arquivos CSV, APIs ou brokers de mensagens.
- **Descrição dos Dados**: Análise da quantidade de linhas, colunas, tipos de dados (numérico contínuo, discreto, categórico).
- **Exploração dos Dados (EDA)**: Avaliação de distribuições estatísticas, médias, medianas e matrizes de correlação.
- **Verificação da Qualidade dos Dados**: Identificação de dados faltantes (*missing values*), erros de digitação, valores discrepantes (*outliers*) e desbalanceamento entre classes.

---

### Fase 3: Data Preparation (Preparação dos Dados)
Esta fase consome entre **70% e 80% do tempo total** de um projeto. Nela, o conjunto de dados bruto é transformado no formato exigido pelos modelos de Machine Learning:
- **Seleção de Dados**: Escolha das tabelas e colunas relevantes; descarte de chaves primárias e números aleatórios.
- **Limpeza de Dados**: Imputação de valores nulos, sanitização de strings e remoção de ruidos.
- **Engenharia de Atributos (Feature Engineering)**: Criação de novas variáveis explicativas calculadas a partir de dados brutos (ex: comprometimento da renda = parcela / renda).
- **Codificação Categórica**: Conversão de textos em colunas numéricas (One-Hot Encoding).

---

### Fase 4: Modeling (Modelagem Preditiva)
Momento de seleção e treinamento dos algoritmos preditivos:
- **Seleção de Algoritmos**: Escolha dos modelos adequados ao problema (ex: Random Forest, XGBoost, Regressão Logística).
- **Estratégia de Validação**: Divisão rigorosa em conjuntos de Treino (75%) e Teste (25%) com amostragem estratificada.
- **Ajuste de Hiperparâmetros**: Configuração dos parâmetros do modelo e ajuste de peso de classes (`class_weight='balanced'`).

---

### Fase 5: Evaluation (Avaliação Técnica e Financeira)
Antes de implantar o modelo em produção, valida-se se as metas técnicas e de negócio foram cumpridas:
- **Avaliação de Métricas**: Análise da Matriz de Confusão, Precision, Recall, F1-Score e área sob a curva ROC (ROC-AUC).
- **Cálculo do ROI Financeiro**: Comparação do prejuízo financeiro com e sem a utilização do modelo preditivo.
- **Auditoria de Feature Importance**: Garantia de que o modelo não está tomando decisões com base em variáveis irrelevantes ou vieses indevidos.

---

### Fase 6: Deployment (Implantação e Operação)
Fase final de operacionalização da solução:
- **Planejamento de Implantação**: Criação de microserviços em Docker e publicação de APIs REST em nuvem.
- **Plano de Monitoramento e Manutenção**: Acompanhamento contínuo da saúde do modelo para identificar a perda de desempenho ao longo do tempo (*Data Drift* e *Concept Drift*).

## 3. Demonstração Prática em Python: Simulação dos Custos Assimétricos de Negócio

No código abaixo, simulamos o cálculo do retorno financeiro gerado por um modelo de Machine Learning aplicado a uma carteira de 1.000 propostas de crédito.


````python
import pandas as pd
import numpy as np

# Simulação da matriz de custos operacionais da Fintech
CUSTO_FALSO_NEGATIVO = 10000.0  # Perda por calote não previsto (FN)
CUSTO_FALSO_POSITIVO = 600.0    # Perda por recusa indevida de bom pagador (FP)

# Cenário de Teste: Carteira com 1.000 proponentes (200 Inadimplentes e 800 Adimplentes)
total_inadimplentes_reais = 200
total_adimplentes_reais = 800

# Cenário A: Aprovação Cega (Sem modelo preditivo - Todos são aprovados)
fn_cenario_a = 200  # Todos os 200 inadimplentes são aprovados (Falsos Negativos)
fp_cenario_a = 0    # Nenhum adimplente é recusado
custo_cenario_a = (fn_cenario_a * CUSTO_FALSO_NEGATIVO) + (fp_cenario_a * CUSTO_FALSO_POSITIVO)

# Cenário B: Modelo de Machine Learning com 85% de Recall de Inadimplência
fn_cenario_b = 30   # Apenas 30 inadimplentes passam sem detecção
fp_cenario_b = 60   # 60 bons pagadores são erroneamente recusados
custo_cenario_b = (fn_cenario_b * CUSTO_FALSO_NEGATIVO) + (fp_cenario_b * CUSTO_FALSO_POSITIVO)

economia_gerada = custo_cenario_a - custo_cenario_b

print("==========================================================")
print("      ANÁLISE DE IMPACTO FINANCEIRO DA FASE 1 (CRISP-DM)  ")
print("==========================================================")
print(f"Custo Operacional Sem Modelo (Aprovação Cega):  R$ {custo_cenario_a:,.2f}")
print(f"Custo Operacional Com Modelo Machine Learning: R$ {custo_cenario_b:,.2f}")
print("----------------------------------------------------------")
print(f"ECONOMIA LÍQUIDA GERADA PELO MODELO:             R$ {economia_gerada:,.2f}")
print("==========================================================")
````
