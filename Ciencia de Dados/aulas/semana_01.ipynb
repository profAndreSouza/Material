## 2. O que é o CRISP-DM?

O **CRISP-DM** foi criado no final dos anos 1990 por um consórcio de grandes empresas para resolver a falta de padronização e a desconexão entre modelos estatísticos e os objetivos reais das empresas.

Diferente do desenvolvimento tradicional de software, o desenvolvimento de modelos preditivos é **experimental, estatístico e orientado a hipóteses**. O CRISP-DM organiza essa incerteza em um mapa claro de 6 fases encadeadas:

1. **Business Understanding (Entendimento do Negócio)**: Mapeamento de problemas, objetivos de negócio, custos assimétricos (Falso Negativo vs Falso Positivo) e critérios de sucesso.
2. **Data Understanding (Entendimento dos Dados)**: Coleta inicial, estatística descritiva, diagnóstico de qualidade de dados e hipóteses.
3. **Data Preparation (Preparação dos Dados)**: Seleção, limpeza, sanitização de outliers, imputação de nulos, feature engineering e codificação.
4. **Modeling (Modelagem Preditiva)**: Algoritmos de Machine Learning, divisão treino/teste e ajuste de hiperparâmetros.
5. **Evaluation (Avaliação)**: Métricas técnicas (Recall, Precision, F1-Score, ROC-AUC) e validação do ROI financeiro.
6. **Deployment (Implantação e Operação)**: Produção via APIs/scripts, monitoramento contínuo e prevenção de data drift.

## 3. Mapeamento de Custos Assimétricos
Em problemas de classificação (por exemplo, aprovação de crédito ou detecção de fraudes), os erros possuem custos totalmente diferentes:
- **Falso Negativo (FN)**: Aprovar um proponente inadimplente (Calote). *Custo alto: perda do principal emprestado.*
- **Falso Positivo (FP)**: Recusar um proponente bom pagador. *Custo baixo: perda de margem de juros.*


```python
# Simulação conceitual do ciclo CRISP-DM e Custos Assimétricos
import pandas as pd
import numpy as np

# Exemplo de matriz de custos em Reais
custo_falso_negativo = 10000.0  # Prejuízo de conceder crédito a inadimplente
custo_falso_positivo = 600.0    # Custo de oportunidade de recusar bom pagador

def calcular_custo_op(fn, fp):
    custo_total = (fn * custo_falso_negativo) + (fp * custo_falso_positivo)
    return custo_total

print(f"Custo de 10 Falsos Negativos: R$ {calcular_custo_op(10, 0):,.2f}")
print(f"Custo de 10 Falsos Positivos: R$ {calcular_custo_op(0, 10):,.2f}")
```
