## FASE 2: Data Understanding (Entendimento dos Dados)

Simularemos a coleta de dados de telemetria dos sensores da esteira (Pressão, Temperatura, Vibração e Rotação) para inspecionar dimensões, tipos de dados e identificar ruídos.

```python
# 2. Simulação / Coleta Inicial de Dados Industriais (Data Understanding)
np.random.seed(42)
n_amostras = 1000

# TODO: Simule as variáveis de sensores usando np.random.normal:
# pressao (loc=6.0, scale=0.8), temperatura (loc=65.0, scale=8.0), vibracao (loc=2.5, scale=0.5), rotacao_rpm (loc=1750, scale=50)
pressao = np.random.normal(loc=6.0, scale=0.8, size=n_amostras)
temperatura = np.random.normal(loc=65.0, scale=8.0, size=n_amostras)
vibracao = np.random.normal(loc=2.5, scale=0.5, size=n_amostras)
rotacao_rpm = np.random.normal(loc=1750, scale=50, size=n_amostras)

# Inserindo ruídos simulados
pressao[10:25] = np.nan
temperatura[50] = 999.0 # Outlier extremo

# Regra física para simular a variável alvo 'falha'
falha = ((pressao > 7.5) | (temperatura > 80.0) | (vibracao > 3.8)).astype(int)

# TODO: Monte o DataFrame df_raw com as colunas: pressao_bar, temperatura_c, vibracao_mms, rotacao_rpm, falha_maquina
df_raw = pd.DataFrame({
    'pressao_bar': pressao,
    'temperatura_c': temperatura,
    'vibracao_mms': vibracao,
    'rotacao_rpm': rotacao_rpm,
    'falha_maquina': falha
})

print("--- Inspeção das Primeiras Linhas --- ")
display(df_raw.head())

print("\n--- Verificação de Valores Nulos --- ")
# TODO: Imprima a contagem de nulos por coluna usando df_raw.isnull().sum()
```
