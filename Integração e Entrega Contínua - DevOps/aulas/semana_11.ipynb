## 2. Prática — Orquestrador da Esteira DevSecOps End-to-End em Python

Nesta atividade prática final, executaremos um script Python que simula todas as 7 etapas da esteira integrada, verificando a aprovação de cada portão de qualidade antes da promoção final para a instância AWS EC2.

````python
import time

def simular_esteira_robusta_devsecops(branch_origem="release/v2.0.0"):
    etapas = [
        ("1. Validação GitFlow & Trigger", True, "Push em 'release/v2.0.0' acionou o workflow no GitHub Actions"),
        ("2. Análise SAST (Bandit)", True, "0 segredos expostos, 0 vulnerabilidades de alta severidade"),
        ("3. Testes Unitários e Integração (Pytest)", True, "24/24 testes passaram | Cobertura: 89.2% (Gate ok)"),
        ("4. Docker Multi-Stage Build & Push", True, "Imagem 'smartn1/app:v2.0.0' publicada no DOCKER HUB"),
        ("5. Provisionamento IaC (Terraform)", True, "Estado da AWS EC2 e Security Group validados sem drift"),
        ("6. Deploy AWS EC2 & Healthcheck Probe", True, "Container atualizado na EC2 | Healthcheck HTTP 200 OK!"),
        ("7. Telemetria & Observabilidade", True, "Métricas RED registradas no Prometheus & Grafana")
    ]
    
    print(f"=== INICIANDO EXECUÇÃO DA ESTEIRA DEVSECOPS END-TO-END ({branch_origem}) ===\n")
    
    for nome, sucesso, detalhe in etapas:
        time.sleep(0.2)
        status_str = "[PASS]" if sucesso else "[FAIL]"
        print(f"{status_str} {nome}")
        print(f"       -> {detalhe}")
        if not sucesso:
            print(f"\n[ALERTA DE SEGURANÇA/QUALIDADE] Falha na etapa '{nome}'. Disparando Rollback...")
            return False
            
    print("\n=========================================================================")
    print("🚀 ESTEIRA DEVSECOPS EXECUTADA COM 100% DE SUCESSO EM PRODUÇÃO NA AWS! 🚀")
    print("=========================================================================")
    return True

simular_esteira_robusta_devsecops()
````
