## 1. Fundamentação Teórica

### 1.1 A Evolução Histórica e a Ruptura dos Silos Organizacionais

Historicamente, a engenharia de software sofria de uma separação rígida entre quem escrevia o código e quem o mantinha operando em servidores de produção:

```text
  +-----------------------------------+               +-----------------------------------+
  |    EQUIPE DE DESENVOLVIMENTO      |               |       EQUIPE DE OPERAÇÕES         |
  |                                   |  MURO DE      |                                   |
  | - Foco: Entregar funcionalidades  |  CONFUSÃO     | - Foco: Estabilidade total        |
  | - Métrica: Quantidade de commits  | ============= | - Métrica: Zero indisponibilidade |
  | - Visão: "No meu PC funciona!"    | (Wall of      | - Visão: "Mudanças causam quedas!"|
  +-----------------------------------+  Confusion)   +-----------------------------------+
                    \                                       /
                     +-------> CONFLITO DE INTERESSES <----+
```

Essa desconexão gerava ciclos de release lentos (trimestrais ou anuais), deploys manuais de alto risco realizados de madrugada, longos períodos de inatividade e empurramento de culpa (*blamestorming*).

O **DevOps** não é um cargo, uma ferramenta ou um departamento separado, mas sim uma **filosofia cultural e de engenharia** baseada na responsabilidade compartilhada sobre o ciclo de vida completo do software — da primeira linha de código ao monitoramento em tempo real no chão de fábrica.

---

### 1.2 O Framework CALMS

Proposto por Jez Humble e ampliado pela comunidade, o framework **CALMS** norteia a transformação cultural e operacional:

![Cultura DevOps e Framework CALMS](img/cultura_devops_calms.jpg)

#### Detalhamento dos 5 Pilares CALMS:

1. **C - Culture (Cultura):**
   - Foco na colaboração contínua, empatia e transparência.
   - Adoção de análises pós-incidente sem culpa (*Post-Mortem Blameless*), onde falhas no sistema são investigadas para melhorar os processos automatizados e não para punir indivíduos.

2. **A - Automation (Automação):**
   - Automação de compilação, testes unitários e de integração, análise de segurança, provisionamento de infraestrutura e publicação.
   - Eliminação de tarefas repetitivas manuais com baixo valor agregado (*Toil*).

3. **L - Lean (Enxuto):**
   - Princípios da manufatura enxuta aplicados ao software: redução do Trabalho em Progresso (*Work In Progress - WIP*), entregas em lotes pequenos e eliminação de gargalos.

4. **M - Measurement (Medição):**
   - Decisões guiadas por dados empíricos reais: tempo de ciclo, métricas de aplicação (latência, erros) e telemetria de negócios.

5. **S - Sharing (Compartilhamento):**
   - Transparência total de conhecimento, ferramentas, erros e sucessos entre times de produto, infraestrutura e segurança.

### 1.3 As 4 Métricas DORA (DevOps Research and Assessment)

O grupo de pesquisa **DORA** (adquirido pelo Google Cloud) estabeleceu quatro métricas padronizadas mundialmente para mensurar a maturidade de engenharia de software:

![Dashboard de Métricas DORA](img/metricas_dora.jpg)

#### Quadro de Desempenho DORA:

| Métrica DORA | Descrição | Desempenho Elite | Desempenho Baixo |
| :--- | :--- | :--- | :--- |
| **Deployment Frequency (DF)** | Frequência de implantação em produção. | Múltiplos deploys por dia | 1 vez a cada 1 a 6 meses |
| **Lead Time for Changes (LTC)** | Tempo do commit mesclado até a execução em produção. | Menos de 1 hora | Entre 1 e 6 meses |
| **Change Failure Rate (CFR)** | Percentual de deploys que exigem correção urgente ou rollback. | $0\% \text{ a } 15\%$ | $46\% \text{ a } 60\%$ |
| **Time to Restore Service (MTTR)** | Tempo médio para recuperar a aplicação após indisponibilidade. | Menos de 1 hora | Mais de 1 semana |

### 1.4 Mapeamento do Fluxo de Valor (Value Stream Mapping - VSM)

O **Value Stream Mapping (VSM)** identifica todas as etapas necessárias para levar uma alteração de código da ideia à produção, separando o **Process Time** (tempo em que o trabalho é efetivamente realizado) do **Lead Time** (tempo total decorrido, incluindo esperas e filas de aprovação):

$$\text{Eficiência do Fluxo de Valor (}\%\text{)} = \frac{\sum \text{Process Time}}{\text{Lead Time Total}} \times 100\%$$

```text
  [Desenvolvimento] -> [Fila de Code Review] -> [Code Review] -> [Fila de QA] -> [Testes QA] -> [Deploy Prod]
   Proc: 4h            Espera: 24h               Proc: 1h         Espera: 48h      Proc: 2h       Proc: 0.5h
```

No exemplo acima:
- $\sum \text{Process Time} = 4\text{h} + 1\text{h} + 2\text{h} + 0.5\text{h} = 7.5\text{ horas}$
- $\text{Lead Time Total} = 4 + 24 + 1 + 48 + 2 + 0.5 = 79.5\text{ horas}$
- $\text{Eficiência do Fluxo} = \frac{7.5}{79.5} \times 100\% \approx 9.4\%$

O objetivo primário do DevOps é automatizar as esperas manuais em filas para elevar essa eficiência a valores superiores a $60\%$.

## 2. Prática — Calculador de VSM e Diagnóstico DORA em Python

Nesta atividade prática, executaremos um script Python completo que analisa um histórico de deploys e filas de processo de um microsserviço fabril da Smart N1, fornecendo o diagnóstico de eficiência do fluxo de valor e o perfil de maturidade DORA.

````python
import datetime

# Histórico de eventos de deploys de microsserviços da Smart N1
historico_deploys = [
    {"id": "DEP-101", "commit": "2026-08-01T08:00:00", "deploy": "2026-08-01T08:35:00", "sucesso": True, "tempo_recup_min": 0, "proc_min": 30, "espera_min": 5},
    {"id": "DEP-102", "commit": "2026-08-02T09:10:00", "deploy": "2026-08-02T09:40:00", "sucesso": True, "tempo_recup_min": 0, "proc_min": 25, "espera_min": 5},
    {"id": "DEP-103", "commit": "2026-08-04T14:00:00", "deploy": "2026-08-04T14:50:00", "sucesso": False, "tempo_recup_min": 20, "proc_min": 35, "espera_min": 15},
    {"id": "DEP-104", "commit": "2026-08-06T10:00:00", "deploy": "2026-08-06T10:25:00", "sucesso": True, "tempo_recup_min": 0, "proc_min": 20, "espera_min": 5},
    {"id": "DEP-105", "commit": "2026-08-08T11:00:00", "deploy": "2026-08-08T11:30:00", "sucesso": True, "tempo_recup_min": 0, "proc_min": 25, "espera_min": 5},
    {"id": "DEP-106", "commit": "2026-08-10T15:00:00", "deploy": "2026-08-10T15:40:00", "sucesso": True, "tempo_recup_min": 0, "proc_min": 35, "espera_min": 5},
]

def diagnosticar_desempenho_devops(deploys, dias_janela=10):
    total_deploys = len(deploys)
    df_diaria = total_deploys / dias_janela
    
    lead_times = []
    falhas = 0
    recuperacoes = []
    proc_times = []
    espera_times = []
    
    fmt = "%Y-%m-%dT%H:%M:%S"
    for dep in deploys:
        dt_commit = datetime.datetime.strptime(dep["commit"], fmt)
        dt_deploy = datetime.datetime.strptime(dep["deploy"], fmt)
        lt_min = (dt_deploy - dt_commit).total_seconds() / 60.0
        lead_times.append(lt_min)
        
        proc_times.append(dep["proc_min"])
        espera_times.append(dep["espera_min"])
        
        if not dep["sucesso"]:
            falhas += 1
            recuperacoes.append(dep["tempo_recup_min"])
            
    lt_medio = sum(lead_times) / len(lead_times)
    cfr_pct = (falhas / total_deploys) * 100.0
    mttr_medio = sum(recuperacoes) / len(recuperacoes) if recuperacoes else 0.0
    
    total_proc = sum(proc_times)
    total_lead = sum(lead_times)
    eficiencia_vsm_pct = (total_proc / total_lead) * 100.0
    
    # Classificação DORA
    perfil = "ELITE" if (df_diaria >= 0.5 and lt_medio <= 60 and cfr_pct <= 15) else "ALTO/MÉDIO"
    
    return {
        "Frequência de Deploy (deploys/dia)": round(df_diaria, 2),
        "Lead Time Médio (minutos)": round(lt_medio, 1),
        "Change Failure Rate (%)": f"{round(cfr_pct, 1)}%",
        "MTTR Médio (minutos)": round(mttr_medio, 1),
        "Eficiência do Fluxo (VSM %)": f"{round(eficiencia_vsm_pct, 1)}%",
        "Nível de Maturidade DORA": perfil
    }

diagnostico = diagnosticar_desempenho_devops(historico_deploys)

print("=========================================================")
print("DIAGNÓSTICO DE DESEMPENHO DEVOPS & VSM — SMART N1")
print("=========================================================\n")
for k, v in diagnostico.items():
    print(f"- {k:35s}: {v}")
````
