## Exercício 2 · Conversão de Unidades e Calculadora EIP-1559 em Python

> **Contexto para Iniciantes:**  
> Assim como 1 Real é divisível em 100 Centavos, 1 Ether é divisível em $1.000.000.000.000.000.000$ (10¹⁸) **Wei**. Para cotar taxas de Gas sem lidar com bilhões ou frações decimais minúsculas, utiliza-se o **Gwei** (1 Gwei = 10⁹ Wei = 0,000000001 ETH).  
> A taxa total pós-EIP-1559 é dada por:  
> $$\text{Taxa Total (Wei)} = \text{Gas Used} \times (\text{Base Fee} + \text{Priority Fee})$$

### 👨‍💻 Sua Tarefa em Código:
Complete a função `calcular_custo_transacao` abaixo para calcular o custo total em ETH e o valor queimado. Em seguida, execute a célula para testar seus resultados com dois cenários reais.

````python
# Exercício 2: Calculadora de Taxas e Conversor de Unidades
def calcular_custo_transacao(gas_used: int, base_fee_gwei: float, priority_fee_gwei: float):
    '''
    Calcula o custo total de uma transação Ethereum e a parcela queimada pelo protocolo.
    
    Retorna um dicionário com:
    - custo_total_eth: float
    - queimado_eth: float
    - recompensa_validador_eth: float
    '''
    # DICA DE CONVERSÃO:
    # 1 Gwei = 10**9 Wei
    # 1 ETH  = 10**18 Wei
    
    preco_efetivo_gwei = base_fee_gwei + priority_fee_gwei
    
    # 1. Calcule o custo total em Wei
    # SEU CÓDIGO AQUI:
    custo_total_wei = gas_used * preco_efetivo_gwei * (10**9)
    
    # 2. Calcule a parcela queimada (Base Fee) em Wei
    # SEU CÓDIGO AQUI:
    queimado_wei = gas_used * base_fee_gwei * (10**9)
    
    # 3. Calcule a recompensa do validador (Priority Fee) em Wei
    # SEU CÓDIGO AQUI:
    validador_wei = gas_used * priority_fee_gwei * (10**9)
    
    # 4. Converta os resultados para ETH dividindo por 10**18
    custo_total_eth = custo_total_wei / (10**18)
    queimado_eth = queimado_wei / (10**18)
    validador_eth = validador_wei / (10**18)
    
    return {
        "custo_total_eth": custo_total_eth,
        "queimado_eth": queimado_eth,
        "validador_eth": validador_eth
    }

# ==============================================================================
# TESTE DOS CENÁRIOS
# ==============================================================================
# Cenário A: Transferência Simples entre duas pessoas (21.000 Gas)
res_a = calcular_custo_transacao(gas_used=21000, base_fee_gwei=25.0, priority_fee_gwei=1.5)
print("--- CENÁRIO A: Transferência Simples ---")
print(f"Custo Total: {res_a['custo_total_eth']:.6f} ETH")
print(f"Queimado:    {res_a['queimado_eth']:.6f} ETH")
print(f"Validador:   {res_a['validador_eth']:.6f} ETH")

# Cenário B: Troca de Moedas em uma Corretora Descentralizada / Uniswap (180.000 Gas)
res_b = calcular_custo_transacao(gas_used=180000, base_fee_gwei=40.0, priority_fee_gwei=2.0)
print("\n--- CENÁRIO B: Swap em DEX (Smart Contract) ---")
print(f"Custo Total: {res_b['custo_total_eth']:.6f} ETH")
print(f"Queimado:    {res_b['queimado_eth']:.6f} ETH")
print(f"Validador:   {res_b['validador_eth']:.6f} ETH")
````

## Exercício 5 · O Guardião do Estado: Implementando a Proteção Anti-Replay com `nonce`

> **Contexto para Iniciantes:**  
> No modelo de contas da Ethereum, se Alice assina digitalmente uma transação enviando 10 ETH para Bob, um invasor que intercepte essa mensagem na rede poderia simplesmente reenviar o mesmo pacote assinado 5 vezes seguidas para esvaziar a conta de Alice (o chamado **Ataque de Reexecução / *Replay Attack***).  
> Para impedir isso, cada conta possui um campo obrigatório chamado **`nonce`**: um número inteiro que começa em 0 e deve crescer de 1 em 1 a cada transação enviada.

### 👨‍💻 Sua Tarefa em Código:
Complete a lógica da função `processar_transacao` abaixo para validar se o `nonce` informado bate exatamente com o `nonce` esperado pela conta. Caso contrário, a transação deve ser sumariamente rejeitada!

````python
# Exercício 5: Validador de Nonce contra Replay Attacks

# Banco de dados em memória simulando contas Ethereum
banco_contas = {
    "0xAlice": {"nonce": 3, "saldo_eth": 50.0},
    "0xBob":   {"nonce": 0, "saldo_eth": 10.0}
}

def processar_transacao(remetente: str, destinatario: str, valor: float, nonce_transacao: int):
    conta = banco_contas.get(remetente)
    
    if not conta:
        return "❌ Conta remetente não encontrada."
        
    print(f"\nTentando processar envio de {valor} ETH de {remetente} para {destinatario}...")
    print(f"Nonce informado no pacote: {nonce_transacao} | Nonce esperado pela rede: {conta['nonce']}")
    
    # 1. VALIDAÇÃO DE NONCE:
    # Verifique se o nonce_transacao é estritamente igual ao conta['nonce']
    # SEU CÓDIGO AQUI:
    if nonce_transacao != conta["nonce"]:
        return f"🚨 BLOQUEADO: Nonce Inválido! Esperado {conta['nonce']}, recebido {nonce_transacao}. Possível Replay Attack!"
        
    # 2. VALIDAÇÃO DE SALDO:
    if conta["saldo_eth"] < valor:
        return "❌ BLOQUEADO: Saldo insuficiente."
        
    # 3. ATUALIZAÇÃO ATÔMICA:
    # Debita do remetente, credita no destinatário e incrementa o nonce do remetente
    conta["saldo_eth"] -= valor
    conta["nonce"] += 1
    banco_contas[destinatario]["saldo_eth"] += valor
    
    return f"✅ SUCESSO! Transação processada. Próximo nonce esperado de {remetente}: {conta['nonce']}"

# ==============================================================================
# TESTES DE SEGURANÇA
# ==============================================================================
# Teste 1: Transação legítima (Alice usa o nonce correto = 3)
msg1 = processar_transacao("0xAlice", "0xBob", 5.0, nonce_transacao=3)
print(msg1)

# Teste 2: O invasor interceptou a transação 1 e tenta reenviá-la com o mesmo nonce 3 (Replay Attack)
msg2 = processar_transacao("0xAlice", "0xBob", 5.0, nonce_transacao=3)
print(msg2)

# Teste 3: Alice envia uma nova transação legítima com o nonce incrementado = 4
msg3 = processar_transacao("0xAlice", "0xBob", 2.0, nonce_transacao=4)
print(msg3)
````
