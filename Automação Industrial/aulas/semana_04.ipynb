## 1. Fundamentação teórica

### 1.1 O Protocolo MQTT na Indústria 4.0
O MQTT é um protocolo leve de transporte de mensagens projetado para dispositivos com restrições de largura de banda em ambientes IIoT. Ele substitui a arquitetura cliente-servidor por um broker desacoplado em tempo e espaço.

---


## 2. Arquitetura da atividade

**Sensores Fabris (Publishers) → Broker MQTT (Mosquitto / HiveMQ) → Clientes de Monitoramento (Subscribers)**

## 3. Prática — Simulação de Cliente MQTT em Python

Execute a simulação de publicação de mensagens de telemetria industrial:

```python
import json
import time
import random

def publicar_telemetria_mqtt():
    topico = "smartn1/esteira01/sensor_temperatura"
    payload = {
        "timestamp": time.strftime("%Y-%m-%d %H:%M:%S"),
        "sensor": "TEMP_MOT_01",
        "valor_celsius": round(random.uniform(45.0, 68.0), 2),
        "qos": 1
    }
    return topico, payload

print("=== SIMULADOR DE PUBLICAÇÃO MQTT SMART N1 ===\n")
topico, payload = publicar_telemetria_mqtt()
print(f"Tópico MQTT: {topico}")
print(f"Payload JSON: {json.dumps(payload, indent=2)}")
```
