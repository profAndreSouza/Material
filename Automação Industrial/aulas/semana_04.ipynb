## 2. Fundamentação Teórica

### 2.1 O Protocolo MQTT
O MQTT é um protocolo leve de transporte de mensagens projetado para dispositivos com restrições de largura de banda e bateria em ambientes IIoT. Ele substitui a arquitetura cliente-servidor tradicional por um broker desacoplado em tempo e espaço.

## 3. Passo a Passo Prático para o Estudante

Executar o código Python de simulação de publicação de telemetria industrial via MQTT.

```python
# Simulacao de Cliente MQTT em Python
import json
import time
import random

def publicar_telemetria_mqtt():
    topico = "smartn1/esteira01/sensor_temperatura"
    payload = {
        "timestamp": time.strftime("%Y-%m-%d %H:%M:%S"),
        "sensor": "TEMP_MOT_01",
        "valor_celsius": round(random.uniform(45.0, 68.0), 2),
        "qos": 1
    }
    return topico, payload

print("=== SIMULADOR DE PUBLICAÇÃO MQTT SMART N1 ===\n")
topico, payload = publicar_telemetria_mqtt()
print(f"Topico MQTT: {topico}")
print(f"Payload JSON: {json.dumps(payload, indent=2)}")
```
