## 3. Passo a Passo Prático para o Estudante

Simulação do pipeline integrado de automação em Python.

```python
# Simulacao Integrada End-to-End de Automacao Industrial
import json
import time
import random

def simular_pipeline_completo():
    # 1. Leitura do Sensor no Chao de Fabrica
    temp_lida = round(random.uniform(40.0, 75.0), 1)
    
    # 2. Publicacao MQTT
    payload_mqtt = {"sensor_id": "TEMP_MOT_01", "valor": temp_lida, "ts": time.time()}
    
    # 3. Tratamento Node-RED e InfluxDB Line Protocol
    status = "ALERTA" if temp_lida > 65.0 else "OK"
    line_protocol = f"telemetria,sensor={payload_mqtt['sensor_id']} temperatura={temp_lida},status=\"{status}\""
    
    return payload_mqtt, line_protocol, status

print("=== SIMULAÇÃO DO PIPELINE INTEGRADO FABRIL ===\n")
for i in range(3):
    mqtt_data, influx_line, status = simular_pipeline_completo()
    print(f"Ciclo {i+1}:")
    print(f"  - Payload MQTT: {json.dumps(mqtt_data)}")
    print(f"  - Line Protocol InfluxDB: {influx_line}")
    print(f"  - Diagnostico: {status}\n")
```
