## 1. Fundamentação teórica

### 1.1 Detecção de Anomalias em Telemetria
Anomalias são padrões nos dados que não se conformam com o comportamento normal do equipamento. Na indústria, desvios de vibração e temperatura antecedem falhas graves.

---


## 2. Arquitetura da atividade

**Matriz de Telemetria (Temperatura e Vibração) → Algoritmo Isolation Forest → Diagnóstico de Anomalias (-1 / 1)**

---


## 3. Prática — Treinamento do Isolation Forest em Python

Execute o algoritmo de detecção de anomalias:

```python
import numpy as np
from sklearn.ensemble import IsolationForest

X = np.array([
    [40.0, 1.2], [41.2, 1.1], [39.5, 1.3], [40.8, 1.2],
    [41.0, 1.0], [92.0, 8.5] # Ponto anômalo
])

clf = IsolationForest(contamination=0.15, random_state=42)
predicoes = clf.fit_predict(X)

print("=== RESULTADO DA DETECÇÃO DE ANOMALIAS ===")
for i, p in enumerate(X):
    status = "ANOMALIA DETECTADA" if predicoes[i] == -1 else "Normal"
    print(f"Leitura {i+1}: Temp = {p[0]} °C, Vib = {p[1]} mm/s | Status: {status}")
```
