### 2.1 Detecção de Anomalias em Telemetria
Anomalias são padrões nos dados que não se conformam com a noção definida de comportamento normal. Na indústria, anomalias de vibração e temperatura antecedem quebras mecânicas.

## 3. Passo a Passo Prático para o Estudante

Treinando o Isolation Forest em Python.

```python
# Algoritmo Isolation Forest em Python
import numpy as np
from sklearn.ensemble import IsolationForest

X = np.array([
    [40.0, 1.2], [41.2, 1.1], [39.5, 1.3], [40.8, 1.2],
    [41.0, 1.0], [92.0, 8.5] # Ponto anomalo
])

clf = IsolationForest(contamination=0.15, random_state=42)
predicoes = clf.fit_predict(X) # -1 indica anomalia, 1 indica normal

print("=== RESULTADO DA DETECÇÃO DE ANOMALIAS ===")
for i, p in enumerate(X):
    status = "ANOMALIA DETECTADA" if predicoes[i] == -1 else "Normal"
    print(f"Leitura {i+1}: Temp = {p[0]} °C, Vib = {p[1]} mm/s | Status: {status}")
```
