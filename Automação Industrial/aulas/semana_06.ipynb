### 2.1 InfluxDB e Séries Temporais
Bancos de dados de séries temporais são otimizados para armazenar dados indexados pelo tempo, permitindo alta taxa de gravação e retenção automatizada de dados históricos industriais.

## 3. Passo a Passo Prático para o Estudante

Simulação de gravação de dados em InfluxDB Line Protocol via Python.

```python
# Gerador de InfluxDB Line Protocol
import time

def formatar_line_protocol(measurement, tags, fields):
    str_tags = ",".join([f"{k}={v}" for k, v in tags.items()])
    str_fields = ",".join([f"{k}={v}" for k, v in fields.items()])
    timestamp_ns = int(time.time() * 1e9)
    return f"{measurement},{str_tags} {str_fields} {timestamp_ns}"

line = formatar_line_protocol(
    measurement="telemetria_esteira",
    tags={"estacao": "EST_01", "sensor": "VIB_01"},
    fields={"vibração_mm_s": 1.45, "temperatura_c": 42.8}
)
print("Formato Line Protocol do InfluxDB:")
print(line)
```
