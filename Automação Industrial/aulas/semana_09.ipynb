## 1. Visão Geral e Objetivos
Compreender a aplicação de algoritmos preditivos para estimativa do tempo restante de vida útil de equipamentos (RUL) e cálculo automatizado do índice de eficiência **OEE (Overall Equipment Effectiveness)**.

### Objetivos de Aprendizagem:
- Entender os pilares do OEE: Disponibilidade, Desempenho e Qualidade.
- Calcular o OEE a partir de dados de telemetria em Python.

## 2. Fundamentação Teórica

### 2.1 Cálculo do OEE (Overall Equipment Effectiveness)
$$\text{OEE} = \text{Disponibilidade} \times \text{Desempenho} \times \text{Qualidade}$$

- **Disponibilidade:** Tempo de Operação Real / Tempo de Operação Planejado.
- **Desempenho:** (Peças Produzidas / Tempo Operacional) / Capacidade Nominal.
- **Qualidade:** Peças Aprovadas / Total de Peças Produzidas.

## 3. Passo a Passo Prático para o Estudante

Cálculo automatizado do OEE da fábrica Smart N1 em Python.

```python
# Calculadora de OEE Industrial em Python
def calcular_oee(tempo_planejado_h, tempo_paradas_h, pecas_totais, pecas_refugadas, taxa_nominal_pecas_h):
    tempo_operacao_h = tempo_planejado_h - tempo_paradas_h
    disponibilidade = tempo_operacao_h / tempo_planejado_h
    
    capacidade_teorica = tempo_operacao_h * taxa_nominal_pecas_h
    desempenho = pecas_totais / capacidade_teorica
    
    pecas_boas = pecas_totais - pecas_refugadas
    qualidade = pecas_boas / pecas_totais
    
    oee = disponibilidade * desempenho * qualidade
    return disponibilidade, desempenho, qualidade, oee

disp, des, qual, oee = calcular_oee(
    tempo_planejado_h=8.0,
    tempo_paradas_h=0.5,
    pecas_totais=2200,
    pecas_refugadas=45,
    taxa_nominal_pecas_h=300
)

print("=== INDICADORES DE EFICIÊNCIA SMART N1 ===")
print(f"Disponibilidade: {disp*100:.2f}%")
print(f"Desempenho:      {des*100:.2f}%")
print(f"Qualidade:       {qual*100:.2f}%")
print(f"OEE Final:       {oee*100:.2f}%")
```
