## 2. Fundamentação Teórica

### 2.1 AWS Lambda e Amazon API Gateway
- **AWS Lambda:** Execução orientada a eventos sem necessidade de gerenciar servidores ou sistemas operacionais.
- **Amazon API Gateway:** Roteador de solicitações HTTP REST seguro e escalável.
- **Amazon DynamoDB:** Banco de dados NoSQL de chave-valor com latência de milissegundos.

## 3. Passo a Passo Prático para o Estudante (Projeto 07)

Simulação de uma função AWS Lambda em Python registrando um alerta de parada de máquina no DynamoDB.

```python
# Simulacao do Handler da Funcao AWS Lambda
import json
import time
import uuid

def lambda_handler(event, context):
    body = json.loads(event.get("body", "{}"))
    alerta_id = str(uuid.uuid4())[:8]
    
    item_dynamodb = {
        "AlertaId": alerta_id,
        "maquina": body.get("maquina", "MAQ_01"),
        "severidade": body.get("severidade", "ALTA"),
        "descricao": body.get("descricao", "Temperatura elevada no motor"),
        "timestamp": time.strftime("%Y-%m-%d %H:%M:%S")
    }
    
    return {
        "statusCode": 201,
        "body": json.dumps({"mensagem": "Alerta registrado no DynamoDB", "item": item_dynamodb})
    }

# Teste de execucao do Handler
evento_requisicao = {"body": json.dumps({"maquina": "PRENSA_02", "severidade": "CRITICA", "descricao": "Pressao insuficiente"})}
resposta = lambda_handler(evento_requisicao, None)
print("Resposta do Lambda:")
print(json.dumps(resposta, indent=2))
```
