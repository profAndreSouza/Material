## 1. Fundamentação teórica

### 1.1 AWS Lambda e Amazon API Gateway
- **AWS Lambda:** execução de código orientada a eventos sem necessidade de provisionar ou gerenciar servidores.
- **Amazon API Gateway:** gerenciador de APIs HTTP REST altamente disponível e seguro.
- **Amazon DynamoDB:** banco de dados NoSQL de chave-valor com resposta em milissegundos.

---


## 2. Arquitetura da atividade

**Requisição HTTP POST → Amazon API Gateway → AWS Lambda (com LabRole) → Tabela Amazon DynamoDB**

## 3. Prática — Simulação de Handler AWS Lambda em Python

Execute a simulação do handler Python gravando alertas de parada no DynamoDB:

```python
import json
import time
import uuid

def lambda_handler(event, context):
    body = json.loads(event.get("body", "{}"))
    alerta_id = str(uuid.uuid4())[:8]
    
    item_dynamodb = {
        "AlertaId": alerta_id,
        "maquina": body.get("maquina", "MAQ_01"),
        "severidade": body.get("severidade", "ALTA"),
        "descricao": body.get("descricao", "Temperatura elevada no motor"),
        "timestamp": time.strftime("%Y-%m-%d %H:%M:%S")
    }
    
    return {
        "statusCode": 201,
        "body": json.dumps({"mensagem": "Alerta registrado no DynamoDB", "item": item_dynamodb})
    }

# Teste do Handler
evento_requisicao = {"body": json.dumps({"maquina": "PRENSA_02", "severidade": "CRITICA", "descricao": "Pressao insuficiente"})}
resposta = lambda_handler(evento_requisicao, None)
print("Resposta da Função AWS Lambda:")
print(json.dumps(resposta, indent=2))
```
