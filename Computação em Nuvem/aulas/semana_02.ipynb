## 2. Fundamentação Teórica

### 2.1 Anatomia da Infraestrutura Global da AWS

![Diagrama Infraestrutura Global AWS](images/aws_infraestrutura_global.jpg)

A AWS constrói sua infraestrutura mundial em uma arquitetura hierárquica e altamente redundante:

1. **Regiões AWS (AWS Regions):**
   - Uma Região é uma área geográfica isolada no mundo contendo múltiplos Data Centers agrupados em Zonas de Disponibilidade.
   - Exemplos de códigos de região: `us-east-1` (Norte da Virgínia, EUA - região principal com maior variedade de serviços e menor custo), `sa-east-1` (São Paulo, Brasil - ideal para menor latência na América do Sul).
   - Nenhuma Região compartilha dados com outra a menos que o usuário configure explicitamente essa replicação.

2. **Zonas de Disponibilidade (Availability Zones - AZs):**
   - Cada Região é composta por no mínimo 3 Zonas de Disponibilidade independentes (ex: `us-east-1a`, `us-east-1b`, `us-east-1c`).
   - Uma AZ consiste em um ou mais Data Centers físicos com energia, refrigeração e segurança física dedicados.
   - As AZs de uma mesma Região são conectadas por redes de fibra óptica de altíssima velocidade e baixíssima latência (< 2ms).

3. **Pontos de Presença (Edge Locations):**
   - Redes de distribuição de conteúdo espalhadas em centenas de cidades pelo mundo para armazenar conteúdo em cache próximo aos usuários (Amazon CloudFront) e acelerar resoluções de DNS (AWS Route 53).

### 2.2 O Ambiente AWS Academy Learner Lab Sandbox

![Fluxograma Sequencial do AWS Learner Lab](images/aws_learner_lab_fluxo.jpg)

O **AWS Academy Learner Lab** fornece uma conta AWS Sandbox real para o desenvolvimento das atividades práticas. É fundamental compreender as regras de operação do ambiente:

- **Orçamento Educacional:** Cada aluno possui um limite pré-pago (geralmente US$ 100). O consumo é contabilizado em tempo real.
- **Tempo de Sessão (4 Horas):** Cada sessão ativa do lab dura no máximo 4 horas. Quando a sessão expira ou o aluno clica em *'End Lab'*, o sistema executa automaticamente a ação `Stop` nas instâncias EC2 para economizar créditos.
- **Persistência dos Recursos:** O desligamento da sessão **NÃO deleta os seus trabalhos**. Volumes de disco EBS, arquivos no Amazon S3, tabelas do DynamoDB e a configuração da VPC **permanecem intactos**. Ao iniciar uma nova sessão (*'Start Lab'*), você apenas religa a instância EC2.

### 2.3 Modelo de Segurança IAM e o Papel da `LabRole`

![Diagrama de Segurança IAM LabRole](images/aws_labrole_permissoes.jpg)

No Learner Lab, a AWS desabilita as permissões de criação de novas roles arbitrárias (`iam:CreateRole`). Para permitir a comunicação entre serviços de forma segura, o laboratório já fornece uma IAM Role pré-criada chamada **`LabRole`** (e seu perfil de instância associado **`LabInstanceProfile`**).

**Como utilizar:** Sempre que uma atividade exigir permissões de acesso (ex: uma EC2 acessando o S3, ou uma função Lambda escrevendo no DynamoDB), você deve selecionar a **`LabRole`** no campo de IAM Role.

## 3. Passo a Passo Prático para o Estudante

Siga o guia passo a passo para inicializar seu ambiente de laboratório e validar o acesso via AWS CLI.

### Passo 1: Inicializar a Sessão do Laboratório
1. Acesse o portal do **AWS Academy** e entre na disciplina de Computação em Nuvem.
2. Clique no menu **AWS Academy Learner Lab**.
3. Clique no botão **Start Lab** no canto superior direito.
4. Aguarde até que o círculo indicador passe de amarelo (starting) para **verde (ready)**.

### Passo 2: Acessar o Console de Gerenciamento da AWS
1. Quando o indicador estiver verde, clique no link **AWS** ao lado do ponto verde.
2. Uma nova aba será aberta com o **AWS Management Console**.
3. No canto superior direito do Console AWS, verifique se a Região selecionada é **N. Virginia (`us-east-1`)**.

### Passo 3: Obter as Credenciais Temporárias da AWS CLI
1. Na tela do Learner Lab (aba do AWS Academy), clique no botão **AWS Details**.
2. Ao lado de *AWS CLI*, clique em **Show**.
3. Copie o bloco de texto que contém `aws_access_key_id`, `aws_secret_access_key` e `aws_session_token`.

### Passo 4: Formatar e Validar as Credenciais via Python
Execute o script Python abaixo para validar a estrutura das credenciais temporárias do seu ambiente.

```python
# Script de Validação e Formatação de Credenciais AWS Learner Lab
import os

def validar_ambiente_aws():
    access_key = os.getenv("AWS_ACCESS_KEY_ID", "EXEMPLO_ACCESS_KEY_LEARNER_LAB")
    secret_key = os.getenv("AWS_SECRET_ACCESS_KEY", "EXEMPLO_SECRET_KEY_LEARNER_LAB")
    session_token = os.getenv("AWS_SESSION_TOKEN", "EXEMPLO_SESSION_TOKEN_LEARNER_LAB")
    regiao_padrao = os.getenv("AWS_DEFAULT_REGION", "us-east-1")
    
    print("=== CHECKLIST DE CREDENCIAIS DO AWS LEARNER LAB ===\n")
    print(f"1. Regiao Alvo: {regiao_padrao}")
    print(f"2. Access Key ID: {access_key[:10]}... (Status: OK)")
    print(f"3. Secret Access Key: {secret_key[:10]}... (Status: OK)")
    print(f"4. Session Token Temporario: {session_token[:15]}... (Status: OK)\n")
    
    print("Comandos para exportar no terminal Linux/macOS:")
    print(f"export AWS_ACCESS_KEY_ID='{access_key}'")
    print(f"export AWS_SECRET_ACCESS_KEY='{secret_key}'")
    print(f"export AWS_SESSION_TOKEN='{session_token}'")
    print(f"export AWS_DEFAULT_REGION='{regiao_padrao}'\n")
    
    print("Comandos para exportar no PowerShell Windows:")
    print(f"$env:AWS_ACCESS_KEY_ID='{access_key}'")
    print(f"$env:AWS_SECRET_ACCESS_KEY='{secret_key}'")
    print(f"$env:AWS_SESSION_TOKEN='{session_token}'")
    print(f"$env:AWS_DEFAULT_REGION='{regiao_padrao}'")

validar_ambiente_aws()
```
