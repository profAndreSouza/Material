## 1. Fundamentação teórica

### 1.1 Infraestrutura global da AWS

![Diagrama Infraestrutura Global AWS](images/aws_infraestrutura_global.jpg)

A infraestrutura global da AWS é construída em uma arquitetura redundante de 3 camadas:

1. **Regiões AWS (AWS Regions):** áreas geográficas isoladas contendo múltiplos Data Centers agrupados em Zonas de Disponibilidade (ex: `us-east-1` N. Virgínia, `sa-east-1` São Paulo).
2. **Zonas de Disponibilidade (Availability Zones - AZs):** instalações físicas de Data Centers isolados dentro de uma Região, conectados por redes de fibra óptica de altíssima velocidade (< 2ms).
3. **Pontos de Presença (Edge Locations):** pontos de borda da rede AWS para cache de conteúdo (Amazon CloudFront) e resolução DNS (AWS Route 53).


### 1.2 O ambiente AWS Academy Learner Lab Sandbox

![Fluxograma Sequencial do AWS Learner Lab](images/aws_learner_lab_fluxo.jpg)

No **AWS Academy Learner Lab**, cada estudante possui um ambiente sandbox isolado:

- **Orçamento limite:** orçamento pré-pago em créditos AWS (ex: US$ 100).
- **Tempo de sessão:** limite de 4 horas por sessão ativa. Ao expirar ou clicar em *End Lab*, o sistema executa automaticamente o `Stop` nas instâncias EC2.
- **Persistência de recursos:** o encerramento da sessão **NÃO deleta os dados**. Discos EBS, buckets S3 e tabelas DynamoDB permanecem salvos.


### 1.3 Modelo de segurança IAM e a `LabRole`

![Diagrama de Segurança IAM LabRole](images/aws_labrole_permissoes.jpg)

No Learner Lab, a criação de novas IAM Roles (`iam:CreateRole`) é restrita. O ambiente disponibiliza a **`LabRole`** (e o **`LabInstanceProfile`**), pré-configurados com permissões para operarmos serviços como EC2, S3, Lambda, DynamoDB e SageMaker com segurança.


## 2. Arquitetura da atividade

O fluxo operacional do Learner Lab segue a sequência:

**Start Lab → Obtenção de Credenciais Temporárias → Acesso ao Console AWS / CLI (`us-east-1`) → Operação com LabRole → End Lab (Stop EC2)**


## 3. Prática — Operação do AWS Learner Lab e Validação de Credenciais

### Passo 1 — Inicializar a Sessão
1. Acesse o portal **AWS Academy**.
2. Clique em **AWS Academy Learner Lab**.
3. Clique em **Start Lab** e aguarde o indicador ficar **verde (ready)**.

### Passo 2 — Acessar o Console AWS
1. Clique no link **AWS** ao lado do indicador verde.
2. No Console AWS, selecione a Região **N. Virginia (`us-east-1`)**.

### Passo 3 — Obter Credenciais Temporárias
1. Clique em **AWS Details → AWS CLI → Show**.
2. Copie as credenciais temporárias do seu laboratório.


### Passo 4 — Validar as Credenciais via Python

Execute o código Python abaixo para checar o status das variáveis de ambiente de acesso AWS.


```python
import os

def validar_ambiente_aws():
    access_key = os.getenv("AWS_ACCESS_KEY_ID", "EXEMPLO_ACCESS_KEY_LEARNER_LAB")
    secret_key = os.getenv("AWS_SECRET_ACCESS_KEY", "EXEMPLO_SECRET_KEY_LEARNER_LAB")
    session_token = os.getenv("AWS_SESSION_TOKEN", "EXEMPLO_SESSION_TOKEN_LEARNER_LAB")
    regiao_padrao = os.getenv("AWS_DEFAULT_REGION", "us-east-1")
    
    print("=== CHECKLIST DE CREDENCIAIS DO AWS LEARNER LAB ===\n")
    print(f"1. Região Alvo: {regiao_padrao}")
    print(f"2. Access Key ID: {access_key[:10]}... (Status: OK)")
    print(f"3. Secret Access Key: {secret_key[:10]}... (Status: OK)")
    print(f"4. Session Token Temporário: {session_token[:15]}... (Status: OK)\n")
    
    print("Comandos para exportar no terminal Linux/macOS:")
    print(f"export AWS_ACCESS_KEY_ID='{access_key}'")
    print(f"export AWS_SECRET_ACCESS_KEY='{secret_key}'")
    print(f"export AWS_SESSION_TOKEN='{session_token}'")
    print(f"export AWS_DEFAULT_REGION='{regiao_padrao}'\n")
    
    print("Comandos para exportar no PowerShell Windows:")
    print(f"$env:AWS_ACCESS_KEY_ID='{access_key}'")
    print(f"$env:AWS_SECRET_ACCESS_KEY='{secret_key}'")
    print(f"$env:AWS_SESSION_TOKEN='{session_token}'")
    print(f"$env:AWS_DEFAULT_REGION='{regiao_padrao}'")

validar_ambiente_aws()
```
