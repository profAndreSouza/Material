## 2. Conteúdo Teórico

### 2.1 Componentes da Infraestrutura Global AWS
- **Regiões AWS:** Locais geográficos no mundo contendo múltiplos Data Centers isolados (ex: `us-east-1` em N. Virginia, `sa-east-1` em São Paulo).
- **Zonas de Disponibilidade (AZs):** Instalações de Data Centers fisicamente distintas dentro de uma Região, interconectadas por redes de fibra óptica de altíssima velocidade e baixíssima latência.
- **Pontos de Presença (Edge Locations):** Locais onde a AWS executa serviços de borda como o Amazon CloudFront (CDN) e AWS Route 53 (DNS).

### 2.3 Estratégia de Permissões com a `LabRole`

No AWS Learner Lab, os alunos não possuem permissão para criar novas IAM Roles (`iam:CreateRole`). Para contornar essa restrição, a AWS disponibiliza a **`LabRole`** (e o `LabInstanceProfile`), que é uma role pré-configurada com permissões para operar serviços como EC2, S3, Lambda, DynamoDB e SageMaker.

## 3. Atividade Prática: Manipulação de Credenciais AWS via Código

No código a seguir, implementamos um manipulador em Python que valida a presença das variáveis de ambiente de acesso da AWS (`AWS_ACCESS_KEY_ID`, `AWS_SECRET_ACCESS_KEY` e `AWS_SESSION_TOKEN`) exigidas em ambientes de linha de comando como o AWS CLI.

```python
# Script de validação e formatação de credenciais temporárias do AWS Learner Lab
import os

def verificar_credenciais_aws():
    access_key = os.getenv("AWS_ACCESS_KEY_ID", "MOCK_ACCESS_KEY_LEARNER_LAB")
    secret_key = os.getenv("AWS_SECRET_ACCESS_KEY", "MOCK_SECRET_KEY_LEARNER_LAB")
    session_token = os.getenv("AWS_SESSION_TOKEN", "MOCK_SESSION_TOKEN_LEARNER_LAB")
    regiao = os.getenv("AWS_DEFAULT_REGION", "us-east-1")
    
    print("=== STATUS DAS CREDENCIAIS AWS NO AMBIENTE ===")
    print(f"Região Padrão: {regiao}")
    print(f"AWS Access Key ID: {access_key[:8]}... (Tamanho: {len(access_key)})")
    print(f"AWS Secret Access Key: {secret_key[:8]}... (Tamanho: {len(secret_key)})")
    print(f"AWS Session Token Presente: {'SIM' if len(session_token) > 0 else 'NÃO'}")
    print("\nPara configurar no terminal Linux/macOS:")
    print(f"export AWS_ACCESS_KEY_ID='{access_key}'")
    print(f"export AWS_SECRET_ACCESS_KEY='{secret_key}'")
    print(f"export AWS_SESSION_TOKEN='{session_token}'")
    print(f"export AWS_DEFAULT_REGION='{regiao}'")

verificar_credenciais_aws()
```
