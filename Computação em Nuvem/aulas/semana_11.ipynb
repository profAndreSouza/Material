## 3. Passo a Passo Prático para o Estudante (Projeto 10)

Simulação de treinamento de um modelo de manutenção preditiva em Python e inferência.

```python
# Treinamento de um Modelo de Machine Learning (Random Forest) para Manutencao Preditiva
import random
from sklearn.ensemble import RandomForestClassifier

# Dataset Simulado de Telemetria Industrial
# Recursos: [Temperatura (°C), Vibracao (mm/s)] | Rotulo: 0 (Normal), 1 (Anomalia/Falha)
X_train = [
    [40.5, 1.1], [42.0, 1.2], [39.8, 0.9], [45.1, 1.5],
    [78.5, 4.8], [82.1, 5.2], [75.0, 4.5], [89.3, 6.1]
]
y_train = [0, 0, 0, 0, 1, 1, 1, 1]

# Treinamento do Modelo
modelo = RandomForestClassifier(n_estimators=10, random_state=42)
modelo.fit(X_train, y_train)

# Teste de Predicao em Tempo Real
nova_leitura_sensor = [[80.2, 5.0]] # Temperatura alta e vibracao elevada
predicao = modelo.predict(nova_leitura_sensor)
probabilidade = modelo.predict_proba(nova_leitura_sensor)

print("=== REQUISIÇÃO DE INFERÊNCIA AO MODELO PREDITIVO SMART N1 ===")
print(f"Leitura Recebida: Temperatura = {nova_leitura_sensor[0][0]} °C | Vibracao = {nova_leitura_sensor[0][1]} mm/s")
print(f"Diagnostico: {'ALERTA DE ANOMALIA / MANUTENÇÃO REQUERIDA' if predicao[0] == 1 else 'OPERACAO NORMAL'}")
print(f"Probabilidade de Falha: {probabilidade[0][1]*100:.1f}%")
```
