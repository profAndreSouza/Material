## 2. Fundamentação Teórica

### 2.1 Amazon VPC (Virtual Private Cloud)
- **VPC:** Rede privada virtual dedicada à sua conta AWS.
- **Subrede Pública:** Possui rota direta para o Internet Gateway (`0.0.0.0/0 -> igw-xxx`). Destinada a servidores Web e Load Balancers.
- **Subrede Privada:** Isolada da internet pública. Destinada a bancos de dados e sistemas de controle industrial (SCADA/CLPs).
- **Internet Gateway (IGW):** Roteador que conecta a VPC à internet pública.
