## 1. Fundamentação teórica

### 1.1 Amazon VPC e subredes
- **VPC:** rede privada virtual dedicada na nuvem AWS.
- **Subrede Pública:** rota direta para a internet via Internet Gateway (`0.0.0.0/0 -> igw-xxx`). Destinada a servidores web.
- **Subrede Privada:** isolada da internet pública. Destinada a bancos de dados e sistemas de controle industrial (SCADA/CLPs).


## 2. Arquitetura da atividade

**Internet → Internet Gateway (IGW) → Tabela de Roteamento → Subrede Pública (VPC) → Subrede Privada (Bancos / CLPs)**
