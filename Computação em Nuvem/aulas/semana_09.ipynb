## 1. Fundamentação teórica

### 1.1 AWS SAM (Serverless Application Model)
O AWS SAM é uma extensão do AWS CloudFormation projetada para simplificar a definição e implantação de stacks Serverless (Lambda, API Gateway, DynamoDB) via modelos YAML sintéticos.

---


## 2. Arquitetura da atividade

**Desenvolvedor (`template.yaml`) → AWS SAM CLI → AWS CloudFormation Stack → Provisionamento Automático (Lambda + DynamoDB)**

---


## 3. Prática — Modelo SAM Declarativo em YAML

Execute o código Python para criar o arquivo `template.yaml`:

```python
template_sam = """
AWSTemplateFormatVersion: '2010-09-09'
Transform: AWS::Serverless-2016-10-31
Description: Stack Serverless Fabrica Smart N1

Resources:
  TabelaAlertas:
    Type: AWS::Serverless::SimpleTable
    Properties:
      TableName: AlertasIndustriais
      PrimaryKey:
        Name: AlertaId
        Type: String

  FuncaoAlerta:
    Type: AWS::Serverless::Function
    Properties:
      Handler: app.lambda_handler
      Runtime: python3.9
      Role: arn:aws:iam::123456789012:role/LabRole
      Events:
        CriarAlerta:
          Type: Api
          Properties:
            Path: /alertas
            Method: post
"""

with open("template.yaml", "w") as f:
    f.write(template_sam)

print("Arquivo template.yaml do AWS SAM gerado com sucesso!")
```
