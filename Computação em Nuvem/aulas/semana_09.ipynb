## 2. Fundamentação Teórica

### 2.1 AWS SAM (Serverless Application Model)
O AWS SAM é uma extensão do AWS CloudFormation projetada para simplificar a definição e implantação de aplicações serverless na AWS via arquivos YAML declarativos.

## 3. Passo a Passo Prático para o Estudante (Projeto 09)

Exemplo de template YAML declarativo do AWS SAM (`template.yaml`).

```python
# Modelo de Template SAM para deploy da Funcao Lambda e Tabela DynamoDB
template_sam = """
AWSTemplateFormatVersion: '2010-09-09'
Transform: AWS::Serverless-2016-10-31
Description: Stack Serverless Fabrica Smart N1

Resources:
  TabelaAlertas:
    Type: AWS::Serverless::SimpleTable
    Properties:
      TableName: AlertasIndustriais
      PrimaryKey:
        Name: AlertaId
        Type: String

  FuncaoAlerta:
    Type: AWS::Serverless::Function
    Properties:
      Handler: app.lambda_handler
      Runtime: python3.9
      Role: arn:aws:iam::123456789012:role/LabRole
      Events:
        CriarAlerta:
          Type: Api
          Properties:
            Path: /alertas
            Method: post
"""

with open("template.yaml", "w") as f:
    f.write(template_sam)

print("Arquivo template.yaml do AWS SAM gerado com sucesso!")
```
