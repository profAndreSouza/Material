## 2. Conteúdo Teórico

### 2.1 Conceitos Fundamentais
- **Elasticidade:** Capacidade de alocar ou desalocar recursos computacionais dinamicamente de acordo com a demanda da planta industrial.
- **Resiliência:** Capacidade de recuperar sistemas automaticamente em caso de falha de hardware ou rede.
- **Escalabilidade Horizontal vs. Vertical:**
  - *Vertical (Scale-up):* Aumento de CPU/RAM em um único servidor.
  - *Horizontal (Scale-out):* Adição de novas instâncias em paralelo com balanceamento de carga.
- **Modelos de Implantação:** Nuvem Pública, Privada e Híbrida.

### 2.2 Modelos de Serviço em Nuvem e a Pirâmide de Automação Industrial

Na manufatura avançada, cada modelo de serviço desempenha um papel na sustentação do ecossistema fabril:

1. **IaaS (Infraestrutura como Serviço):** Provisionamento de máquinas virtuais, armazenamento e redes (ex: instâncias Amazon EC2 para servidores de aplicação/banco de dados).
2. **PaaS (Plataforma como Serviço):** Ambientes gerenciados para desenvolvimento e deploy sem gestão do sistema operacional (ex: AWS Elastic Beanstalk, serviços de banco gerenciados).
3. **SaaS (Software como Serviço):** Softwares finalizados consumidos via Web/API (ex: sistemas ERP, plataformas de analytics fabril).

## 3. Atividade Prática: Simulação de Comparação de Custo e Disponibilidade

Abaixo executamos um script Python simulando o cálculo de SLA (Availability) de um sistema de automação industrial hospedado On-Premises vs. em Nuvem AWS com Redundância.

```python
# Script de simulação de disponibilidade (SLA) para infraestrutura fabril

def calcular_downtime_anual(sla_porcentagem):
    horas_ano = 365.25 * 24
    horas_downtime = horas_ano * (1 - (sla_porcentagem / 100))
    minutos_downtime = horas_downtime * 60
    return horas_downtime, minutos_downtime

cenarios = {
    "On-Premises (Servidor Único na Fábrica)": 98.5,
    "Nuvem AWS (Instância EC2 Única)": 99.9,
    "Nuvem AWS (Multi-AZ com Balanceamento)": 99.99
}

print("=== SIMULAÇÃO DE SLA E TEMPO DE PARADA NA INDÚSTRIA ===\n")
for nome, sla in cenarios.items():
    h, m = calcular_downtime_anual(sla)
    print(f"Cenário: {nome}")
    print(f"  - SLA Garantido: {sla}%")
    print(f"  - Tempo máximo de parada estimado/ano: {h:.2f} horas ({m:.1f} minutos)\n")
```
