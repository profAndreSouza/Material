## 1. Fundamentação teórica

### 1.1 Os 5 pilares do NIST para Computação em Nuvem

De acordo com a especificação do NIST (Special Publication 800-145), a Computação em Nuvem é caracterizada por 5 pilares fundamentais:

1. **Autoatendimento sob demanda (On-demand self-service):** provisionamento automático de recursos computacionais sem necessidade de intervenção humana com o provedor.
2. **Amplo acesso à rede (Broad network access):** recursos acessíveis por mecanismos padrão através da rede (estações, celulares, CLPs, coletores industriais).
3. **Agrupamento de recursos (Resource pooling):** recursos computacionais compartilhados em modelo multilocatário (*multi-tenant*).
4. **Rápida elasticidade (Rapid elasticity):** capacidade de alocar e liberar recursos rapidamente de acordo com a variação de demanda.
5. **Serviço mensurável (Measured service):** medição transparente e cobrança proporcional ao uso dos recursos (*Pay-As-You-Go*).


### 1.2 CapEx × OpEx e TCO na Indústria

A infraestrutura tradicional **On-Premises** exige alto investimento inicial (**CapEx - Capital Expenditure**) em hardware, nobreaks, sistemas de refrigeração de Data Center e licenças.

A **Computação em Nuvem** substitui esse modelo por despesas operacionais (**OpEx - Operational Expenditure**), reduzindo o Custo Total de Propriedade (**TCO - Total Cost of Ownership**).

| Característica | On-Premises | Nuvem AWS (Pay-As-You-Go) |
|---|---|---|
| Modelo financeiro | CapEx (Investimento inicial alto) | OpEx (Custo proporcional ao uso) |
| Tempo de provisionamento | Semanas ou meses | Segundos ou minutos |
| Manutenção de hardware | Total responsabilidade interna | 100% gerenciada pela AWS |
| Escalabilidade | Limitada pelo hardware físico pré-adquirido | Imediata e sob demanda |


### 1.3 Modelos de serviço na Indústria 4.0

![Infográfico Nuvem Indústria 4.0](images/nuvem_industria_40.jpg)

A integração entre os níveis da pirâmide de automação industrial (ISA-95) e a nuvem ocorre em 3 camadas:

1. **IaaS (Infraestrutura como Serviço):** a AWS fornece hardware, rede e virtualização; a empresa gerencia SO, aplicação e dados (ex: instâncias EC2 rodando servidores de banco de dados ou gateways de comunicação).
2. **PaaS (Plataforma como Serviço):** a AWS gerencia o SO, linguagem e banco de dados; a empresa gerencia apenas o código da aplicação (ex: AWS Lambda ou Beanstalk).
3. **SaaS (Software como Serviço):** software completo fornecido via Web/API (ex: sistemas ERP, BI e plataformas de análise preditiva fabril).

## 3. Prática — Simulação de SLA e Indisponibilidade em Python

Execute o código Python abaixo para calcular o tempo máximo de parada anual (*downtime*) e a estimativa de perda financeira na fábrica Smart N1.


```python
def analisar_disponibilidade():
    horas_ano = 365.25 * 24
    cenarios = [
        {"nome": "On-Premises (Servidor Fabril Sem Redundância)", "sla": 98.5, "custo_hora_parada": 5000.0},
        {"nome": "AWS EC2 Instância Única (IaaS)", "sla": 99.9, "custo_hora_parada": 5000.0},
        {"nome": "AWS Arquitetura Multi-AZ com Load Balancer", "sla": 99.99, "custo_hora_parada": 5000.0}
    ]
    
    print("=== SIMULAÇÃO DE DISPONIBILIDADE E PERDA FINANCEIRA (SMART N1) ===\n")
    for c in cenarios:
        downtime_horas = horas_ano * (1 - (c["sla"] / 100))
        downtime_minutos = downtime_horas * 60
        perda_financeira = downtime_horas * c["custo_hora_parada"]
        
        print(f"Cenário: {c['nome']}")
        print(f"  - SLA Garantido: {c['sla']}%")
        print(f"  - Indisponibilidade Estimada: {downtime_horas:.2f} horas ({downtime_minutos:.1f} minutos)")
        print(f"  - Perda Financeira Estimada/Ano: R$ {perda_financeira:,.2f}\n")

analisar_disponibilidade()
```
