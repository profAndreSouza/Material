## 2. Fundamentação Teórica

### 2.1 Os 5 Pilares Fundamentais do NIST para Computação em Nuvem
De acordo com a definição do NIST (Special Publication 800-145), uma solução só pode ser classificada como Computação em Nuvem se atender a cinco características essenciais:

1. **Autoatendimento sob Demanda (On-Demand Self-Service):** O usuário pode provisionar recursos computacionais (tempo de servidor, armazenamento em rede) de forma automática, sem necessidade de interação humana com o provedor.
2. **Amplo Acesso à Rede (Broad Network Access):** Os recursos estão disponíveis através da rede e são acessados por meio de mecanismos padrões, promovendo o uso por plataformas heterogêneas (dispositivos móveis, estações de trabalho, CLPs, coletores industriais).
3. **Agrupamento de Recursos (Resource Pooling):** Os recursos computacionais do provedor são agrupados para servir a múltiplos consumidores utilizando um modelo multilocatário (multi-tenant), com diferentes recursos físicos e virtuais atribuídos e reatribuídos dinamicamente.
4. **Rápida Elasticidade (Rapid Elasticity):** Capacidades podem ser provisionadas e liberadas rapidamente, em alguns casos automaticamente, para escalar proporcionalmente com a demanda (horizontal e verticalmente).
5. **Serviço Mensurável (Measured Service):** Os sistemas de nuvem controlam e otimizam o uso de recursos utilizando um mecanismo de medição em algum nível de abstração apropriado ao tipo de serviço (ex: armazenamento, processamento, largura de banda e contas de usuário ativas).

### 2.2 Impacto Financeiro: CapEx vs. OpEx e TCO na Indústria

Na infraestrutura tradicional **On-Premises**, a empresa realiza grandes investimentos iniciais de capital (**CapEx - Capital Expenditure**) na aquisição de servidores físicos, nobreaks, sistemas de refrigeração para Data Center, licenças de software e cabeamento estruturado.

Na **Computação em Nuvem**, o modelo financeiro é convertido para despesas operacionais (**OpEx - Operational Expenditure**), onde a empresa paga apenas pelos recursos efetivamente consumidos (modelo Pay-As-You-Go), reduzindo o Custo Total de Propriedade (**TCO - Total Cost of Ownership**).

| Característica | On-Premises (Data Center Próprio) | Nuvem AWS (Pay-As-You-Go) |
| :--- | :--- | :--- |
| Modelo Financeiro | CapEx (Alto investimento inicial) | OpEx (Custo proporcional ao uso) |
| Provisionamento | Semanas ou meses para compra/entrega | Segundos ou minutos via console/API |
| Manutenção Física | Responsabilidade total da equipe interna | 100% gerenciada pela AWS |
| Escalabilidade | Limitada pela capacidade física adquirida | Praticamente ilimitada e imediata |
| Atualização de Hardware | Ciclos de substituição a cada 3-5 anos | Transparente e contínua pelo provedor |

### 2.3 Modelos de Serviço em Nuvem na Indústria 4.0

![Infográfico Nuvem Indústria 4.0](images/nuvem_industria_40.jpg)

A relação entre a pirâmide de automação industrial (ISA-95) e a arquitetura de nuvem opera em três camadas:

1. **IaaS (Infraestrutura como Serviço):**
   - *O que a AWS fornece:* Hardware físico, rede, sistemas de refrigeração, camada de virtualização (KVM/Nitro).
   - *O que o cliente gerencia:* Sistema Operacional, atualizações de segurança, Docker Engine, aplicações e dados.
   - *Exemplo na fábrica:* Instância Amazon EC2 rodando servidores de banco de dados PostgreSQL ou gateways de comunicação.

2. **PaaS (Plataforma como Serviço):**
   - *O que a AWS fornece:* Tudo do IaaS mais a instalação, configuração e manutenção do Sistema Operacional, linguagem de execução e banco de dados.
   - *O que o cliente gerencia:* Apenas o código da aplicação e as configurações do sistema.
   - *Exemplo na fábrica:* AWS Elastic Beanstalk ou AWS Lambda para execução de rotinas de cálculo industrial.

3. **SaaS (Software como Serviço):**
   - *O que a AWS/Parceiro fornece:* O software completo e pronto para uso via navegador ou API.
   - *O que o cliente gerencia:* Configuração de usuários e entrada/visualização de dados.
   - *Exemplo na fábrica:* Painéis de BI, ERP em nuvem ou sistemas de manutenção preditiva prontos.

## 3. Passo a Passo Prático para o Estudante

Neste laboratório teórico-prático, você irá simular o cálculo do impacto de indisponibilidade (Downtime) e a avaliação financeira da infraestrutura da fábrica **Smart N1**.

### Passo 1: Executar a Simulação de Disponibilidade (SLA)
Rode a célula Python abaixo para comparar o tempo máximo de parada anual permitido para sistemas hospedados em servidor único na fábrica vs. arquiteturas na nuvem AWS.

```python
# Codigo de Simulacao de SLA e Tempo Maximo de Parada Anual

def analisar_disponibilidade():
    horas_ano = 365.25 * 24
    cenarios = [
        {"nome": "On-Premises (Servidor Fabril Sem Redundancia)", "sla": 98.0, "custo_hora_parada": 5000.0},
        {"nome": "AWS EC2 Instancia Unica (IaaS)", "sla": 99.9, "custo_hora_parada": 5000.0},
        {"nome": "AWS Arquitetura Multi-AZ com Load Balancer", "sla": 99.99, "custo_hora_parada": 5000.0}
    ]
    
    print("=== SIMULACAO DE DISPONIBILIDADE E PERDA FINANCEIRA (SMART N1) ===\n")
    for c in cenarios:
        downtime_horas = horas_ano * (1 - (c["sla"] / 100))
        downtime_minutos = downtime_horas * 60
        perda_financeira = downtime_horas * c["custo_hora_parada"]
        
        print(f"Cenario: {c['nome']}")
        print(f"  - SLA Garantido: {c['sla']}%")
        print(f"  - Indisponibilidade Estimada: {downtime_horas:.2f} horas ({downtime_minutos:.1f} minutos)")
        print(f"  - Perda Financeira Estimada/Ano: R$ {perda_financeira:,.2f}\n")

analisar_disponibilidade()
```

### Passo 2: Analisar a Pirâmide de Automação e Mapeamento de Cargas

Com base nos resultados acima, analise onde cada componente da planta **Smart N1** deve ser posicionado:
- **CLPs e IHMs (Tempo Real Crítico - Latência < 10ms):** Devem permanecer na borda (Edge / Chão de fábrica).
- **Servidores MES e Historian (Latência aceitável < 100ms):** Hospedados em instâncias **Amazon EC2 (IaaS)** com armazenamento SSD (EBS).
- **ERP e Dashboards Estratégicos:** Consumidos via **SaaS** ou plataformas Web escaláveis.
