## Fundamentação Teórica

### 1. O Roteiro de Implantação em Ondas (Phased Roadmap)
A implantação da Governança de TI não ocorre da noite para o dia. Tentar implementar todas as políticas e frameworks simultaneamente causa paralisia operacional. Por isso, utiliza-se a estratégia de **Implantação em Ondas**:

```text
  ONDA 1: CURTO PRAZO (0 a 6 meses)  ---> "Ganho Rápido (Quick Wins)"
                                           Foco em contensão de danos, inventário, 
                                           controle de acessos e criação do Comitê de TI.

  ONDA 2: MÉDIO PRAZO (6 a 18 meses) ---> "Estruturação & Processos"
                                           Adoção do ITIL/COBIT, formalização do PETI,
                                           implementação de ferramentas de monitoramento.

  ONDA 3: LONGO PRAZO (18 a 36 meses)---> "Otimização & Inovação"
                                           Maturidade CMMI/MPS-SW, automação avançada, 
                                           Governança de IA e BI preditivo.
```

### 2. Fatores Críticos de Sucesso (FCS)
Os FCS são os poucos elementos essenciais que, se atendidos, garantem o sucesso da implantação da Governança de TI:
1. **Patrocínio Executivo Mandatório (Tone at the Top):** O CEO e a Diretoria Executiva devem defender a governança ativamente.
2. **Foco no Negócio (Não na Burocracia):** As políticas de TI devem simplificar os negócios, não paralisar as vendas.
3. **Vitórias Rápidas (Quick Wins):** Demonstrar resultados tangíveis nos primeiros 90 dias para obter credibilidade.
4. **Comunicação Transparente:** Capacitar os colaboradores e explicar o "porquê" de cada nova regra.

### 3. Gestão da Mudança Organizacional (Modelo ADKAR)
A maior barreira para a governança de TI raramente é tecnológica; é **cultural**. O modelo **ADKAR** (Prosci) orienta a transição de pessoas:

- **A - Awareness (Conscientização):** Compreensão sobre a necessidade da mudança.
- **D - Desire (Desejo):** Vontade pessoal de apoiar e participar da mudança.
- **K - Knowledge (Conhecimento):** Saber como mudar (treinamentos e políticas claras).
- **A - Ability (Habilidade):** Capacidade de aplicar o novo comportamento no trabalho diário.
- **R - Reinforcement (Reforço):** Ações para sustentar a mudança e celebrar conquistas.

### 4. Orçamento de TI: CAPEX vs. OPEX
- **CAPEX (Capital Expenditure):** Investimento em bens de capital (compra de servidores físicos, licenças perpétuas, reforma de data center). Entra como ativo no balanço.
- **OPEX (Operational Expenditure):** Despesas operacionais recorrentes (assinatura de nuvem SaaS/AWS, terceirização, suporte contratado). Entra como custo operacional dedutível.
- *Tendência de Governança:* Transição de modelos baseados em CAPEX para OPEX flexível na nuvem (Pay-as-you-go).

## Estudo de Caso / Exemplo Prático

### A Implantação da Governança em uma Cooperativa Agroindustrial
- **O Desafio:** A cooperativa possuía 40 filiais de recebimento de grãos com sistemas descentralizados e sem padrões de segurança.
- **Onda 1 (Quick Win):** Padronização dos antivírus e centralização da gestão de acessos (Active Directory) em 90 dias.
- **Onda 2 (Médio Prazo):** Unificação dos bancos de dados das filiais em nuvem privada e implementação da matriz RACI no suporte.
- **Gestão da Mudança:** Criação do programa "TI Parceira do Campo", oferecendo treinamento prático aos gerentes das filiais antes de cada bloqueio de sistema.
