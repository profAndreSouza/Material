## 1. Qualidade do processo e qualidade do produto

**Processo** é o conjunto repetível de atividades, papéis e controles usados para criar ou manter um sistema. **Produto** é o software entregue e seus componentes. Os dois se relacionam: um processo consistente ajuda a evitar erros recorrentes, mas não garante sozinho que o produto atenda às necessidades.

**Exemplo:** uma loja virtual recebe correções diretamente no servidor, sem registrar mudanças. Uma correção urgente resolve um erro de preço, mas derruba o pagamento. O problema não se limita ao código: faltaram registro, avaliação de impacto, teste e plano de retorno.

| Pergunta | Foco | Evidência possível |
|---|---|---|
| O trabalho é repetível e acompanhado? | Processo | fluxo documentado, responsáveis, registros de revisão |
| O sistema atende às necessidades e restrições? | Produto | critérios de aceite, testes, indicadores e feedback |

> Processo formal demais também pode atrapalhar. O controle deve ser proporcional ao risco, ao tamanho da equipe e ao impacto da falha.


## 2. Maturidade: CMMI e MR-MPS-SW

Modelos de maturidade ajudam a avaliar capacidades e planejar evolução. Eles não são certificados de que um produto específico está livre de defeitos, nem substituem diagnóstico baseado em evidências.

### CMMI

O CMMI organiza práticas de capacidade. Na representação por estágios, os níveis de maturidade são **1 a 5**. Em linguagem resumida:

| Nível | Leitura didática | Exemplo observável |
|---|---|---|
| 1 — Inicial | Resultados dependem de esforço individual e variam muito | só uma pessoa sabe publicar e não há plano de teste |
| 2 — Gerenciado | Projetos são planejados e acompanhados | tarefas, responsáveis, prazos e mudanças registrados por projeto |
| 3 — Definido | Práticas comuns são padronizadas e adaptadas | equipes seguem fluxo organizacional documentado |
| 4 — Gerenciado quantitativamente | Desempenho é controlado com dados quantitativos | limites de tempo de resposta e taxa de falhas monitorados |
| 5 — Em otimização | Dados e análise orientam melhoria contínua | causas recorrentes são eliminadas e resultados reavaliados |

### MR-MPS-SW

O Modelo de Referência MPS para Software, da Softex, propõe **sete níveis de maturidade, de G a A**. A progressão em degraus menores pode apoiar uma evolução gradual, conforme o contexto e os objetivos da organização.

| Nível | Nome | Leitura resumida |
|---|---|---|
| G | Parcialmente Gerenciado | iniciar gestão de projetos e requisitos |
| F | Gerenciado | ampliar controles e processos de apoio |
| E | Parcialmente Definido | definir processos organizacionais e recursos |
| D | Largamente Definido | consolidar processos de engenharia |
| C | Definido | incorporar práticas avançadas e integração |
| B | Gerenciado Quantitativamente | gerir quantitativamente projetos/processos |
| A | Em Otimização | melhorar continuamente com inovação |

Os nomes são uma visão introdutória; consulte o guia oficial da versão adotada para requisitos, processos e resultados esperados.

### Como escolher uma referência

- Considere porte, riscos, clientes, contratos, capacidade de investimento e objetivo de negócio.
- Não prometa equivalência direta entre um nível MPS-SW e um nível CMMI: escopos, requisitos e métodos de avaliação diferem.
- Não busque “subir de nível” por prestígio. Defina primeiro qual problema precisa melhorar e como será comprovado.

**Exemplo:** uma software house pequena com atrasos e retrabalho pode começar por requisitos rastreáveis, planejamento simples, controle de versão e revisão de mudanças. Só depois avalia se faz sentido buscar uma avaliação formal de maturidade.


## 3. ISO/IEC 12207: processos ao longo do ciclo de vida

A ISO/IEC 12207 estabelece processos de ciclo de vida para sistemas e software. Ela oferece uma estrutura para organizar atividades desde aquisição e desenvolvimento até operação, manutenção e descontinuação. É uma referência de processos, não uma receita única de sequência ou ferramenta.

Uma forma simples de visualizar o ciclo:

**Necessidade → requisitos → projeto e construção → verificação/validação → transição → operação e manutenção → retirada**

Atividades de apoio — como garantia da qualidade, gestão de configuração, verificação, validação, revisão e auditoria — podem ocorrer ao longo de várias etapas. A organização adapta os processos ao contrato, produto e contexto.

**Exemplo — aplicativo de agendamento:**
1. A área usuária solicita reduzir faltas em consultas.
2. Requisitos definem confirmação, cancelamento e proteção dos dados pessoais.
3. A equipe desenha e implementa a solução sob controle de versão.
4. Testes verificam regras de agendamento e integração com notificações.
5. Uma mudança é aprovada e implantada primeiro em ambiente controlado.
6. Operação acompanha falhas e tempo de resposta; manutenção corrige defeitos.
7. Se o sistema for substituído, dados e acessos são tratados no encerramento.

### Verificação e validação

- **Verificação:** o produto foi construído conforme os requisitos e especificações? Ex.: testar que o sistema impede dois agendamentos no mesmo horário.
- **Validação:** o produto resolve a necessidade real do usuário? Ex.: confirmar com a recepção que o fluxo reduz faltas sem dificultar o atendimento.

As duas atividades se complementam; passar em testes técnicos não prova por si só que a solução é útil.

## 4. Qualidade do produto: ISO/IEC 9126 e ISO/IEC 25010

A ISO/IEC 9126 é uma referência histórica de qualidade de produto de software. A família SQuaRE sucedeu essa linha; para estudo atual, a **ISO/IEC 25010:2023** define um modelo de qualidade de produto com **nove características**. Por isso, listas didáticas antigas com oito características devem ser identificadas como referentes à edição anterior (ISO/IEC 25010:2011).

| Característica (2023) | Pergunta prática | Exemplo de evidência |
|---|---|---|
| Adequação funcional | Faz corretamente o que os usuários precisam? | testes de regras de negócio e cobertura de requisitos críticos |
| Eficiência de desempenho | Responde e usa recursos adequadamente? | tempo de resposta sob carga, consumo de recursos |
| Compatibilidade | Convive e troca dados com outros sistemas? | testes de integração e interoperabilidade |
| Capacidade de interação | É compreensível e fácil de operar? | taxa de conclusão de tarefa, erros de uso, acessibilidade |
| Confiabilidade | Mantém o serviço e se recupera de falhas? | disponibilidade, falhas, tempo de recuperação |
| Segurança | Protege dados e operações contra ameaças? | testes de autorização, vulnerabilidades corrigidas |
| Manutenibilidade | Pode ser analisado, alterado e testado com segurança? | tempo para corrigir, complexidade, cobertura de testes |
| Flexibilidade | Adapta-se a contextos e mudanças? | esforço para configurar ou migrar ambiente |
| Segurança (safety) | Evita riscos inaceitáveis à vida, saúde ou propriedade? | análise de perigos e controles, quando aplicável |

A edição 2023 distingue segurança da informação (*security*) e segurança operacional (*safety*). Para sistemas comuns, algumas características terão prioridade baixa; para sistemas industriais, médicos ou críticos, *safety* pode ser essencial.

> Os termos em português podem variar conforme tradução adotada. Consulte a edição oficial aplicável ao trabalho acadêmico ou contrato.

### Da característica ao critério mensurável

“Sistema rápido” é vago. Um critério precisa de contexto, método e limite acordado.

| Necessidade | Critério de exemplo | Como medir |
|---|---|---|
| Consultar pedidos sem demora | 95% das consultas abaixo de 2 s com 200 usuários simultâneos | teste de carga antes da publicação |
| Evitar perda de pedidos | nenhuma perda de transação confirmada em teste de recuperação | simulação de falha e conferência do registro |
| Proteger dados pessoais | usuário de atendimento não acessa dados administrativos sem autorização | teste de perfis e revisão de permissões |

Os valores são ilustrativos: cada equipe deve estabelecer metas com base na linha de base, risco e expectativa do negócio.

## 5. Gestão de mudanças e garantia da qualidade

Mudança é qualquer alteração planejada em software, infraestrutura, configuração ou serviço. O controle reduz o risco de alterações não rastreáveis e permite aprender com falhas. O processo deve ser leve para mudanças pequenas e mais rigoroso quando o impacto potencial é alto.

### Fluxo proporcional de mudança

1. **Registrar:** abrir ticket com motivo, escopo, solicitante e urgência.
2. **Avaliar:** analisar impacto em usuários, segurança, dados, integrações, custo e continuidade.
3. **Classificar:** distinguir mudança padrão pré-aprovada, normal ou emergencial, conforme política local.
4. **Aprovar:** obter aprovação da autoridade adequada ao risco. Um CAB (comitê consultivo de mudanças) pode apoiar decisões de maior impacto; não precisa aprovar cada ajuste rotineiro.
5. **Testar:** validar em ambiente apropriado e documentar resultados e critérios de aceite.
6. **Implantar:** seguir janela e plano de comunicação, com responsável identificado.
7. **Reverter se necessário:** definir gatilhos e plano de retorno antes da implantação.
8. **Revisar e encerrar:** registrar resultado, incidentes, evidências e lições aprendidas.

**Exemplo — atualização do cálculo de frete:** ticket ligado à regra solicitada; análise identifica impacto em pedidos e integração com transportadora; equipe testa faixas de CEP e pedidos de referência; responsável aprova; implantação ocorre em janela combinada; se aumentar divergência de cobrança, volta-se à versão anterior e a causa é analisada.

### QA não é apenas “testar no final”

Garantia da qualidade (QA) busca confiança de que processos e resultados atendem aos critérios acordados. Pode incluir revisão de requisitos, padrões de código, revisão por pares, automação de testes, auditoria de registros e análise de defeitos. O teste é uma atividade importante, mas não é sinônimo de QA.

## 6. Estudo de caso: Loja Horizonte

A Loja Horizonte vende pela internet e possui uma equipe de quatro pessoas. Pedidos duplicados aparecem em períodos de promoção; alterações são feitas diretamente na produção; o tempo de resposta piora quando há muitos acessos. Os clientes reclamam de falhas no pagamento.

### Diagnóstico inicial (hipotético para a atividade)

| Evidência observada | Interpretação | Prioridade |
|---|---|---|
| Não há registro consistente de requisitos e mudanças | gestão do trabalho pouco previsível | alta |
| Publicações sem homologação | risco de regressão e interrupção | alta |
| Nenhum teste de carga antes de campanhas | desempenho não é conhecido sob pico | alta |
| A equipe identifica defeitos só por reclamações | falta monitoramento e feedback estruturado | média/alta |

Não é possível atribuir formalmente um nível de maturidade apenas com essa descrição; seria preciso avaliar o escopo, as práticas e as evidências conforme o modelo escolhido. Para fins didáticos, a equipe pode descrever capacidades próximas de um estágio inicial e justificar essa hipótese sem alegar uma avaliação oficial.

### Plano de melhoria em etapas

| Horizonte | Ação | Evidência de conclusão | Indicador ilustrativo |
|---|---|---|---|
| 0–2 meses | versionar código e registrar requisitos prioritários | repositório, histórico de commits e backlog | % de mudanças com ticket associado |
| 2–4 meses | criar homologação e revisão por pares | checklist de publicação e registros de revisão | % de publicações testadas em homologação |
| 4–6 meses | automatizar testes críticos de pedido e pagamento | execução repetível no pipeline | defeitos escapados para produção por versão |
| 6–9 meses | testar carga e monitorar disponibilidade | relatório de carga e painel operacional | p95 de resposta e disponibilidade mensal |

Os prazos e metas são exemplos. A equipe deve ajustá-los à capacidade e à criticidade do negócio. O plano melhora capacidades concretas sem presumir que a organização conquistou um nível formal.

### Priorização de qualidade

Para a Loja Horizonte, as primeiras características podem ser **adequação funcional**, **confiabilidade**, **eficiência de desempenho**, **segurança** e **compatibilidade** com o meio de pagamento. A escolha deve vir do diagnóstico e das consequências de falha, não de uma lista genérica.


## 7. Aplicação no Capítulo 8 do PDGTI

O Capítulo 8 deve traduzir o diagnóstico dos capítulos anteriores em ações, responsáveis, evidências e medidas. Use a tabela abaixo como molde e substitua os exemplos pela realidade da empresa escolhida.

| Problema (Cap. 2) | Capacidade/processo relacionado | Ação proposta | Responsável | Indicador e meta | Prazo |
|---|---|---|---|---|---|
| Publicações causam indisponibilidade | gestão de mudanças e configuração | ticket, homologação, aprovação por risco e plano de retorno | líder de TI | % de mudanças com evidência de teste; meta a definir após baseline | 90 dias |
| Requisitos mudam sem rastreabilidade | requisitos e gestão de projeto | relacionar requisito, tarefa, teste e aceite | analista de negócio | % de requisitos críticos rastreáveis | 60 dias |
| Sistema fica lento em picos | eficiência de desempenho | estabelecer teste de carga e monitoramento | desenvolvimento/operação | p95 de resposta no volume acordado | 120 dias |

### Checklist do Capítulo 8

- [ ] Delimitar o escopo avaliado: equipe, sistema, serviço ou processo.
- [ ] Descrever evidências atuais e limitações do diagnóstico.
- [ ] Escolher CMMI **ou** MR-MPS-SW como referência de maturidade, justificando a escolha; não inventar equivalência.
- [ ] Propor pelo menos três práticas de melhoria, conectadas às dores do Capítulo 2.
- [ ] Selecionar características da ISO/IEC 25010 pertinentes ao produto e definir critérios mensuráveis.
- [ ] Descrever fluxo de mudança proporcional ao risco, incluindo testes e retorno.
- [ ] Indicar responsáveis, prazos, linha de base, metas e forma de acompanhamento.
- [ ] Distinguir plano interno de melhoria de uma avaliação formal de maturidade.

### Exemplo de redação para adaptar

> “Como o diagnóstico identificou publicações sem testes em ambiente de homologação, propõe-se implantar um fluxo de mudança com registro, análise de impacto, testes de aceite e plano de retorno para alterações de alto risco. A equipe de desenvolvimento será responsável pela evidência dos testes; o responsável pelo serviço aprovará a janela de publicação. Nos primeiros 30 dias será levantada a linha de base de falhas relacionadas a mudanças. A meta será definida após essa medição e revisada mensalmente.”
