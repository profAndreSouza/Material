## Governança e gestão
**Governança** define direção, prioridades e como acompanhar resultados. **Gestáo** organiza e executa o trabalho diário.

Imagine uma viagem: governança escolhe o destino e confere o rumo; gestão planeja o caminho e dirige o carro.

## 1. ISO/IEC 38500: orientar e acompanhar

Norma com princípios que ajudam dirigentes a tomar boas decisões sobre o uso da TI. Não  é um manual de tarefas técnicas.

**Exemplo:** a direção avalia que o sistema de vendas cai muito; dirige um plano de melhoria; monitora as quedas e o tempo de recuperação.

Seis princípios: **responsabilidade, estratégia, aquisição, desempenho, conformidade e comportamento humano**.

## 2. COBIT: organizar e acompanhar a gestão de TI

Oferece objetivos e práticas para conectar TI às metas da organização e acompanhar controles. Não  é preciso adotar tudo: escolha conforme necessidades e riscos.

| Sigla | Em poucas palavras | Pergunta simples |
|---|---|---|
| **EDM** | Avaliar, dirigir e monitorar | Estamos tomando boas decisões? |
| **APO** | Alinhar, planejar e organizar | Temos plano e recursos? |
| **BAI** | Construir, adquirir e implementar | Como colocar a mudança em uso? |
| **DSS** | Entregar, servir e suportar | Como manter e apoiar os serviços? |
| **MEA** | Monitorar, avaliar e analisar | Como medir os resultados? |

**Exemplo:** se projetos atrasam e estouram o orçamento, identifique a causa e priorize objetivos de projetos, recursos ou acompanhamento que ajudem a resolvê-la.

## 3. ITIL: melhorar os serviços de TI

ITIL reúne orientações para planejar, entregar e melhorar serviços. Um serviço ajuda alguém a trabalhar, como e-mail, internet ou sistema de vendas.

O **Sistema de Valor de Serviço (SVS)** reúne princípios, governança, práticas e melhoria contínua. Atividades comuns: planejar, melhorar, conversar com pessoas, preparar mudanças e entregar suporte.

**Exemplo:** se alguém não acessa o e-mail, a **gestão de incidentes** busca restaurar o serviço rapidamente. Depois, a equipe pode investigar a causa para evitar repetição.

| Referência | Ajuda principalmente com... | Exemplo |
|---|---|---|
| ISO/IEC 38500 | Decisões e responsabilidades da direção | Definir quem aprova prioridades de TI |
| COBIT | Objetivos, controles e acompanhamento | Acompanhar riscos e projetos importantes |
| ITIL | Serviços e práticas do dia a dia | Registrar incidentes e medir tempo de solução |

Podem ser usadas juntas. Escolha partes úteis e evite processos difíceis de manter.

### Exemplo

| Problema | Escolha | Ação | Como saber se melhorou? |
|---|---|---|---|
| O sistema de vendas cai e ninguém sabe quem responde | ISO: responsabilidades; COBIT: continuidade; ITIL: incidentes | Definir responsáveis, plano de continuidade e registro de incidentes | Menos tempo fora do ar e recuperação mais rápida |

> **Dica:** escreva para a realidade da organização. Uma empresa pequena pode começar com poucas práticas bem aplicadas.

2. COBIT estrutura governança e gestão; ITIL orienta práticas para entregar e melhorar serviços.  
3. Gestáo de incidentes no ITIL e continuidade no COBIT são possibilidades; a escolha depende da causa.
